```python
import pandas as pd
import numpy as np

# Load the dataset
df = pd.read_csv('files/csv/superstore_extract.csv')

# 1. View top/bottom rows
print(df.head(5))
print(df.tail(5))

# 2. Schema breakdown (Equivalent to DESCRIBE table_name; in SQL)
print(df.info())

# 3. Shape of DataFrame (Rows, Columns)
print(df.shape)  # Returns tuple: (2000, 6)

# 4. Statistical Summary (Mean, Min, Max, Quantiles)
print(df.describe())
```

```python
robust_report = df.pivot_table(
    index='category',
    columns='quantity',
    values='sales',
    aggfunc='sum',
    fill_value=0,
    margins=True  # Injects Grand Totals automatically!
)
print(robust_report.head(10))
```

```python
# Single column (returns a Series)
sales_series = df['sales']
print(sales_series)
# Multiple columns (returns a DataFrame)
selected_df = df[['order_id', 'sales', 'category']]
print(selected_df)
```

```python
high_sales = df[df['sales'] > 3000]
print(high_sales)
```

```python
tech_high_sales = df[(df['sales'] > 3000) & (df['category'] == 'Technology')]
print(tech_high_sales)
furniture_or_loss = df[(df['category'] == 'Furniture') | (df['profit'] < 0)]
print(furniture_or_loss)
```

```python
categories = ['Furniture', 'Technology']
filtered_cats = df[df['category'].isin(categories)]
print(filtered_cats)
```

```python
mid_range_sales = df[df['sales'].between(1000, 2000)]
print(mid_range_sales)
```

```python
df = df.rename(columns={
    'sales': 'total_sales',
    'profit': 'net_profit'
})
```

```python
# Aggregations across entire series
total_sales = df['total_sales'].sum()
print(total_sales)
average_profit = df['net_profit'].mean()
print(average_profit)
median_discount = df['discount'].median()
print(median_discount)
max_quantity = df['quantity'].max()
print(max_quantity)
# Frequency counts of unique values (SQL: GROUP BY category COUNT(*))
category_distribution = df['category'].value_counts()
print(category_distribution)
```

```python
# SQL: SELECT category, AVG(total_sales) FROM table GROUP BY category;
avg_sales_by_cat = df.groupby('category')['total_sales'].mean()
print(avg_sales_by_cat)
```

```python
# SQL: 
# SELECT category, SUM(total_sales) AS revenue, AVG(net_profit) AS avg_profit, COUNT(order_id) AS orders 
# FROM table GROUP BY category;

grouped_summary = df.groupby('category').agg(
    revenue=('total_sales', 'sum'),
    avg_profit=('net_profit', 'mean'),
    orders=('order_id', 'count')
).reset_index()

print(grouped_summary)
```

```python
# SQL: SELECT * FROM table GROUP BY category HAVING SUM(total_sales) > 1600000;
large_categories_df = df.groupby('category').filter(lambda x: x['total_sales'].sum() > 1600000)
print(large_categories_df)
```

```python
# SQL: SELECT *, AVG(total_sales) OVER(PARTITION BY category) AS category_avg_sales FROM table;

df['category_avg_sales'] = df.groupby('category')['total_sales'].transform('mean')

# Calculate the contribution percentage of each individual sale to its parent category
df['sale_percentage_of_category'] = (df['total_sales'] / df.groupby('category')['total_sales'].transform('sum')) * 100

print(df[['category', 'total_sales', 'category_avg_sales', 'sale_percentage_of_category']].sort_values(by='category').head(10))
```

```python
# Vectorized multi-conditional logic (SQL: CASE WHEN ... THEN ... ELSE ...)
conditions = [
    (df['net_profit'] > 500),
    (df['net_profit'] >= 0) & (df['net_profit'] <= 500),
    (df['net_profit'] < 0)
]
choices = ['High Profit', 'Stable', 'Loss-Making']

df['performance_tier'] = np.select(conditions, choices, default='Unknown')

# Custom python lambda logic via apply
df['adjusted_quantity'] = df['quantity'].apply(lambda x: x + 1 if x % 2 == 0 else x)
df
```

```python
# SQL: ORDER BY category ASC, total_sales DESC
df_sorted = df.sort_values(by=['category', 'total_sales'], ascending=[True, False])
df_sorted
```

```python
pivot_df = df.pivot_table(
    index='category', 
    values=['total_sales', 'net_profit'], 
    aggfunc={'total_sales': 'sum', 'net_profit': 'mean'}
)
pivot_df
```

```python
# 1. ROW_NUMBER() OVER(PARTITION BY category ORDER BY sales DESC)
df['row_number'] = df.groupby('category')['total_sales'].rank(method='first', ascending=False)
df.sort_values(by=['category','row_number'],ascending=[True, False])
```

```python
# Grouping by multiple dimensions creates a MultiIndex DataFrame
multi_df = df.groupby(['category', 'quantity']).agg({'total_sales': 'mean', 'net_profit': 'sum'})
print(multi_df.head())

# 1. Slicing a MultiIndex using pd.IndexSlice (SQL: Filtering on composite keys)
# Get 'Furniture' records where quantity is between 5 and 7
idx = pd.IndexSlice
furniture_slice = multi_df.loc[idx['Furniture', 5:7], :]
print(furniture_slice)
```

```python
# 2. Flattening the MultiIndex columns (Crucial after running complex .agg features)
# If your agg created nested columns like ('sales', 'mean'), flatten them:
multi_df.columns = ['_'.join(col).strip() for col in multi_df.columns.values]
multi_df = multi_df.reset_index()
print(multi_df)
```

```python
df['sales_cleaned'] = df['total_sales'].mask(df['total_sales'] > 4000, np.nan)
df
```

```python
criteria = [
    (df['total_sales'] > 3000) & (df['net_profit'] > 500),
    (df['total_sales'] > 3000) & (df['net_profit'] <= 500),
    (df['total_sales'] <= 3000)
]
outputs = ['High Yield', 'High Risk Volume', 'Standard']
df['deal_segment'] = np.select(criteria, outputs, default='Unclassified')
df
```

```python
# 1. pd.cut: Equal-width bins (divides the value range into equal parts)
# Divides the min-max sales range into 3 equal spans
df['sales_equal_width'] = pd.cut(df['total_sales'], bins=3, labels=['Low Range', 'Mid Range', 'High Range'])

# 2. pd.qcut: Equal-frequency bins (Sample Quantiles)
# Ensures that an equal number of rows/orders end up in each bucket (Terciles)
df['sales_percentile_bucket'] = pd.qcut(df['total_sales'], q=3, labels=['Tier 3', 'Tier 2', 'Tier 1'])

df
```

```python
# 1. Pandas .where(): Keeps values where condition is TRUE, otherwise replaces with alternative
# SQL: CASE WHEN sales > 1000 THEN sales ELSE 0 END
df['sales_clipped'] = df['total_sales'].where(df['total_sales'] > 1000, 0)

# 2. Pandas .mask(): Replaces values where condition is TRUE (Opposite of .where())
# Hides negative profit records by setting them to NaN
df['clean_profit'] = df['net_profit'].mask(df['net_profit'] < 0, np.nan)

# 3. NumPy np.where(): Binary choice assignment (True choice, False choice)
df['profit_status'] = np.where(df['net_profit'] > 0, 'Profitable', 'Loss')
df
```

```python
target_sequences = range(1000, 1010)

# 2. Aligning two DataFrames with mismatched axes
df1 = df.iloc[0:10, :]
df2 = df.iloc[5:15, :]
aligned_df1, aligned_df2 = df1.align(df2, join='outer', axis=0)
print(aligned_df1)
```

```python
# Create an explicit logical ordering tier
custom_category_order = ['Office Supplies', 'Technology', 'Furniture']

# Convert the column to a CategoricalDtype specifying logical sorting rules
df['category'] = pd.Categorical(df['category'], categories=custom_category_order, ordered=True)

# Now sorting will automatically respect your custom list arrangement instead of alphabetical order
df_custom_sorted = df.sort_values('category')
df_custom_sorted
```
