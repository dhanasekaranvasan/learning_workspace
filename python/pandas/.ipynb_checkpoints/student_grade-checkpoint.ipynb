```python
import pandas as pd
import numpy as np
df = pd.read_csv('../files/csv/student_grades.csv')
df.info()
```

```python
df = df.drop_duplicates('student_id')
df.info()
```

```python
df['total'] = df[['maths','science','english']].sum(axis=1)
df['%'] = df[['maths','science','english']].mean(axis=1)
df['rank'] = df['total'].rank(method='min', ascending=False)
df['rank'] = df['rank'].astype(np.int64)
df.sort_values('rank')
```
