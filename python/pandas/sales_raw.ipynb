```python
import pandas as pd

df = pd.read_csv("../files/csv/sales_raw.csv")
df.info()
```

```python
df.head(10)
```

```python
df['price'] = df['price'].apply(pd.to_numeric, errors='coerce')
cols = ['event_time', 'sale_date']
df[cols] = df[cols].apply(pd.to_datetime, errors='coerce')
df['item'] = df['item'].str.title()
df['status'] = df['status'].str.lower()
df.info()
```

```python
df.head(10)
```

```python
nan = df.groupby(['item', 'status'])['price'].transform('mean')
df['price'] = df['price'].fillna(nan)
df.info()
```

```python
duplicates = df.duplicated(['order_id', 'item', 'price', 'status'])
duplicates.index = df['order_id']
duplicates[duplicates.values == True]
```
