# Pandas DataFrame

## Scalar Series
Series is a 1-Dimentional labeled array

```python
# create scalar series
scalar = pd.Series(5)
print(scalar)
print(scalar[0])
print(scalar.ndim)
print(scalar.dtype)
print(scalar.shape)
print(scalar.index)
print(list(scalar.keys()))
print(list(scalar.values))
```

```python
# scalar arithmetic operations
scalar2 = pd.Series(8)
print(scalar+10)
print(scalar-20)
print(scalar*7)
print(scalar%3)
print(scalar/2)
print(scalar//4)
print(scalar**6)
print(scalar^7)
print(scalar+scalar2)
print(scalar-scalar2)
print(scalar*scalar2)
print(scalar%scalar2)
print(scalar/scalar2)
print(scalar//scalar2)
print(scalar**scalar2)
print(scalar^scalar2)
```

```python
# comparision operations
scalar2 = pd.Series(8)
print(scalar==10)
print(scalar!=20)
print(scalar>7)
print(scalar<3)
print(scalar>=2)
print(scalar<=4)
print(scalar==scalar2)
print(scalar!=scalar2)
print(scalar>scalar2)
print(scalar<scalar2)
print(scalar>=scalar2)
print(scalar<=scalar2)
```

```python
# set name
scalar = pd.Series(10, name='Num1')
scalar2 = pd.Series(8, name='Num2')
print(scalar)
print(scalar2)
```

```python
# set index
scalar = pd.Series(10,index=['A'])
scalar2 = pd.Series(8,index=['B'])
print(scalar)
print(scalar2)
```

```python
# concat scalar series
A = pd.Series(6, index=['A'], name='Num1')
B = pd.Series(8, index=['B'], name='Num2')
scalar = pd.concat((A, B))
print(scalar)
print(scalar['A'])
print(scalar['B'])
print(scalar.ndim)
print(scalar.shape)
print(scalar.dtype)
```

## list and ndarray

```python
A = pd.Series([101, 102, 103, 104, 105, 106, 107, 108, 109, 110])
print(A)
B = pd.Series(np.arange(101, 111))
print(B)
print(B[:])
print(B[:7])
print(B[::-1])
B[5:9]=[206,207,208,209]
print(B)
```

```python
## dict series
series = pd.Series({'Name': 'Dhanasekaran', 'Age': 32, 'Gender': 'Male'})
print(series)
print(series['Name'])
print(series['Age'])
print(series[['Name', 'Gender']])
```

## DataFrame

```python
# create dataframe
df = pd.DataFrame([np.arange(100, 111)])
print(df)
print(df.ndim)
print(df.shape)
```

```python
df = pd.DataFrame(np.arange(100, 111))
print(df)
print(df.ndim)
print(df.shape)
```

```python
names = pd.Series(['Arun', 'Kirthika', 'Banu', 'Kumar', 'Vasanth', 'Venba', 'Chandru'], name='Names')
ages = pd.Series([40,24,27,32,19,36,20])
genders = pd.Series(['M','F','F','M','M','F','M'])
df = pd.DataFrame({'Name': names, 'Age': ages})
df['Gender'] = genders
df
```

```python
df1 = pd.DataFrame({'id': ['A', 'B', 'C', 'D'],
                    'age': [30, 23, 25, 22]})
df2 = pd.DataFrame({'id': ['C', 'D', 'E', 'F'],
                    'job': ['Doctor', 'Statistician',
                            'Accountant', 'Developer']})
df = df1.merge(df2)
df
```
