```python
import pandas as pd
import numpy as np

url = "https://raw.githubusercontent.com/dsrscientist/dataset1/master/titanic_train.csv"
df = pd.read_csv(url)
print(df.info())
print(f'{' shape ':-^100}')
print(df.shape)
print(f'{' sum() ':-^100}')
print(df.sum(numeric_only=True))
print(f'{' count() ':-^100}')
print(df.count())
print(f'{' max() ':-^100}')
print(df.max(numeric_only=True))
print(f'{' min() ':-^100}')
print(df.min(numeric_only=True))
```

```python
ages = df.groupby(['Pclass', 'Sex'])['Age'].transform('median')
df['Age'] = df['Age'].fillna(ages)
df.info()
```

```python
df['Age'] = df['Age'].astype(np.int16)
df['Fare'] = df['Fare'].round(2)
embarked = df.groupby(['Pclass', 'Sex'])['Embarked'].transform(lambda x: x.mode()[0] if not x.mode().empty else 'S')
df['Embarked'] = df['Embarked'].fillna(embarked)
df.info()
```
