```python
# open('file_name', mode)
# mode = r, w, a, x +r, +w
    # x -> create file, error if file exists
    # r -> read file content
    # w -> write file replace all
    # a -> append content to the end
    # b -> binary mode for images
```

```python
# create file (FileExistsError if file already exists)
# write function create new content in a file
file = open('files/test.txt', 'x')
file.write('This is a sample file handing text first line') # write content into file
file.write('\nThis is a sample file handling text second line') # write content into file
file.close()
```

```python
# read file (FileNotFoundError if file not found)
print("---read()---")
file = open('files/test.txt')
print(file.read()) # read all content in a file
file.close()

print("\n---readline()---")
file = open('files/test.txt')
print(file.readline()) # read current cursor line position
print(file.readline()) # read current cursor line position
file.close()

print("\n---readlines()---")
file = open('files/test.txt')
lines = file.readlines() # get all the lines into list
for line in lines:
    print(line)
file.close()

print("\n---readlines()---")
file = open('files/test.txt')
lines = file.readlines() # get all the lines into list
for line in lines:
    print(line)
file.close()

print("\n---readlines()---")
file = open('files/test.txt')
lines = file.readlines() # get all the lines into list
for line in lines:
    print(line)
file.close()
```

```python
# with handling close the file after the operations
with open('files/test.txt') as f:
    print(f.read())
```

```python
# mode append
with open('files/test.txt', 'a') as f:
    f.write('\nThis is a new append line')

with open('files/test.txt') as f:
    print(f.read())
```
