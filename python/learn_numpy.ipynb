```python
import numpy as np
def output(arr):
    print("--------")
    print('array ->',arr)
    print('dim   ->', arr.ndim)
    print('dtype ->', arr.dtype)
    print('shape ->', arr.shape)
    print('size  ->', arr.size)
    print("--------")
```

```python
# create 0D (Scalar) array
arr = np.array(5)
print("Ouput 1:")
output(arr)

arr = np.array((2.5))
print("Ouput 2:")
output(arr)

arr = np.array("Hello")
print("Ouput 3:")
output(arr)

arr = np.array((True))
print("Ouput 4:")
output(arr)
```

```python
# change dtype
arr = np.array(3, dtype=np.int64)
print("Before:", arr.dtype)
arr = arr.astype(np.str_)
print("After:", arr.dtype)
print("-"*50)
arr = np.array("Hello")
print("Before:", arr.dtype)
arr = arr.astype(np.bool_)
print("After:", arr.dtype)
print("-"*50)
arr = np.array(True)
print("Before:", arr.dtype)
arr = arr.astype(np.int64)
print("After:", arr.dtype)
print("-"*50)
arr = np.array("6.2")
print("Before:", arr.dtype)
arr = arr.astype(np.float64)
print("After:", arr.dtype)
```

```python
# arithmetic operation (+, -, *, %, /, //)
# supported dtypes: int, float, uint
print("Arithmetic Operation")
print("-"*30)

arr1 = np.array(5)
arr2 = np.array(2)
arr3 = np.array("Hello")
arr4 = np.array('hello')
print("Addition        :", arr1 + arr2)
print("Subtraction     :", arr1 - arr2)
print("Multipilication :", arr1 * arr2)
print("Modules         :", arr1 % arr2)
print("Divition        :", arr1 / arr2)
print("Floor           :", arr1 // arr2)

# comparison operation (==, !=, >, <, <=, >=)
# 0 == 0.0 == False -> True
# 1 == 1.0 == True  -> True
print("-"*30)
print("Comparison Operation")
print("-"*30)
print(f'{arr3} == {arr4} :', arr3 == arr4)
print(f'{arr3} != {arr4} :', arr3 != arr4)
print(f'{arr1} > {arr2}          :', arr1 > arr2)
print(f'{arr1} < {arr2}          :', arr1 < arr2)
print(f'{arr1} >= {arr2}         :', arr1 >= arr2)
print(f'{arr1} <= {arr2}         :', arr1 <= arr2)
```
